**6 catégories de soins:** 
- Hospitalisation 
- soins_medicaux 
- pharmacie 
- optique 
- dentaire 
- divers --> voir avec Alptis à quoi ça correspond 

A chaque fois, pour chaque catégorie on a des informations sur:  
- ***nbre_decompte*** : le nbre de remboursement réalisé sur des soins : plus précisément le nombre de fois où la Sécurité social a envoyé (ou transmis à la mutuelle) un relevé de remboursement. 
- ***remb_alptis*** : montant remboursement payé par Alptis 
- ***remb_ro*** : remeboursement par le Régime Obligataire = sécurité sociale 
- ***frais_rééls*** : montant frais total engagés (somme remboursement alptis + RO ) 
- ***reste_ a_charge*** : ce que le client doit payer après remboursement RO et Alptis 

On a les montants et le nombre de décompte pour chaque catégorie de soin, mais également le nombre de décompte total et le montant des soins totaux, toute catégories de soin confondues. 

Remarques importantes :  
-  **Remb_alptis+ remb_ro + reste_a_charge= frais_reels**
- **Nb_decomptes = Nb_decomptes_pharmacie + Nb_decomptes_optique + Nb_decomptes_soins_medicaux + Nb_decomptes_hospitalisation + Nb_decomptes_divers**

```python
#transformation de annee_mois_paiement en colonne str
df_consommations["annee_mois_paiement"] = (
    df_consommations["annee_mois_paiement"]
    .astype("string")
    .str.strip()
)

#extraction de toutes les colonnes sauf code_client et annee_mois_paiement
cols_num = df_consommations.select_dtypes(include=["float64"]).columns.tolist()

#Aggrégation en faisant la somme sur toutes les lignes, sur chaque colonne selon la valeur 
#code_client pour les colonnes de cols_num, et en faisant une liste de str pour annee_mois_paiement
df_consommations_agg = df_consommations.groupby("client_code").agg(
    {
        **{col: "sum" for col in cols_num},              # somme robuste aux NaN
        "annee_mois_paiement": lambda x: sorted([v for v in x.dropna().unique()])  # liste sans NaN
    }
).reset_index()
```

```python
#Vérification de la bonne aggrégation
df_agg= df_consommations_agg
print((df_agg["nb_decomptes"] == df_agg["nb_decomptes_dentaire"] + df_agg["nb_decomptes_pharmacie"] + df_agg["nb_decomptes_soins_medicaux"] + df_agg["nb_decomptes_optique"] + + df_agg["nb_decomptes_divers"] + df_agg["nb_decomptes_hospitalisation"]).all())
print(max([c for c in df_agg["client_code"].value_counts()])) #On a bien 1 code client par ligne
```

Nous avons crée le DataFrame *df_consommations_agg*, via la démarche suivante: 
- Pour toutes les colonnes sauf annee_mois_paiement et client_code: on fait la somme sur toutes les lignes d'un même code_client
- Pour annee_mois_paiement: pour chaque ligne code_client, avoir une liste des annee_mois_paiement des remboursements réalisés par alptis. 
Ainsi on conserve l'information concernant la date et la fréquence des remboursements. --> Utile pour analyser si client consulte énormément ou pas et si pic de consultation à un moment dans l'année.

**Remarque:** Le nombre de dates de paiement enregistrées dans l'année pour un code_client particulier (*annee_mois_paiement*) ne correspond pas au nombre de décomptes totaux dans l'année pour ce même code client (*nb_decomptes*). En effet, un client peut recevoir divers remboursements, provenant de soins divers (de la même catégorie ou non) pendant le même mois.


**Problème**:
Cette aggégration nous fait perdre l'information du type de soin et du montant payé (par Alptis, le RO et le client) associé à chaque date de remboursement.
Tentons de la conserver en créant une nouvelle colonne ***info_par_annee_paiement*** dans le DataFrame, avec chaque ligne au format suivant: 

***{anne_mois_paiement: soin : {rb_alptis, rb_ro, reste_a_charge, frais_reels}}*** 

avec soin dans [hospitalisation, dentaire, hospitalisation, optique, soins_medicaux, divers]

```python
#Création d'un dictionnaire mapping_soin pour nous aider dans l'extraction des informations
soins = ["hospitalisation", "soins_medicaux", "pharmacie", "optique", "dentaire", "divers"]
mapping_soins = {
    soin: {
        "frais_reels": f"frais_reels_{soin}",
        "reste_a_charge": f"reste_a_charge_{soin}",
        "remb_alptis": f"remb_alptis_{soin}",
        "remb_ro": f"remb_ro_{soin}",
    }
    for soin in soins
}

#Création d'une fonction qui extrait pour chaque date, le soin associé et les montants associés
def build_info_dict(df_client):
    info = {}

    for _, row in df_client.iterrows():
        date = row["annee_mois_paiement"]

        if date not in info:
            info[date] = {}

        for soin, cols in mapping_soins.items():

            # Récupérer les 4 montants
            fr = row[cols["frais_reels"]]
            rac = row[cols["reste_a_charge"]]
            alpt = row[cols["remb_alptis"]]
            ro = row[cols["remb_ro"]]

            # Vérifier si le soin est présent (= au moins une valeur non nulle)
            if (fr != 0) or (rac != 0) or (alpt != 0) or (ro != 0):
                info[date][soin] = {
                    "frais_reels": fr,
                    "reste_a_charge": rac,
                    "remb_alptis": alpt,
                    "remb_ro": ro,
                }

        # Si aucun soin présent → supprimer la date vide
        if info[date] == {}:
            del info[date]

    return info
```

```python
df_agg["consommation_info_par_annee_mois_paiement"] = (
    df_consommations.groupby("client_code")
    .apply(build_info_dict)
    .reset_index(drop=True)
)
df_consommations_agg=df_agg
```
